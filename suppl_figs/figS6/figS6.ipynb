```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
plt.style.use('default')
mpl.rcParams['axes.linewidth'] = 5  # set the value globally
mpl.rcParams['xtick.major.size'] = 20
mpl.rcParams['xtick.major.width'] = 7
mpl.rcParams['xtick.minor.size'] = 10
mpl.rcParams['xtick.minor.width'] = 7
mpl.rcParams['ytick.major.size'] = 20
mpl.rcParams['ytick.major.width'] = 7
mpl.rcParams['ytick.labelsize'] = 50
mpl.rcParams['xtick.labelsize'] = 50
mpl.rcParams['ytick.minor.size'] = 10
mpl.rcParams['ytick.minor.width'] = 7
mpl.rcParams['font.size'] = 55
mpl.rcParams['font.sans-serif'] = 'Arial'
mpl.rcParams['figure.dpi'] = 500
mpl.rcParams['svg.fonttype'] = 'none'
```

```python
df = pd.read_pickle(r"timelineplot_infection.pkl")
```

```python
from natsort import natsort_keygen 
delta = "\u0394"
col = 4

fig, ax = plt.subplots(col, int(np.floor(len(df['construct'].unique())/col)), figsize=(30,35), sharex=True, sharey=True)
ax =ax.flatten()

colors = {
    'Growth': 'k',
    'noninfected': 'dodgerblue',
    'infected': 'purple',
}

construct_group = df.groupby(['construct'], as_index=False, observed=False)

for index, tup in enumerate(construct_group):
    cur_ax = ax[index]
    current_df = tup[1]
    experiment_parameter_group = current_df.groupby(['experimentparameter'], observed=False)
    for sec_ind, tup2 in enumerate(experiment_parameter_group):
        new_plotting_df = tup2[1]
        parameter = tup2[0][0]
        color = colors[parameter]
        new_plotting_df = new_plotting_df.sort_values(by=['timepoint'], key=natsort_keygen())
        
        cur_ax.scatter(new_plotting_df['timepoint'], new_plotting_df['biological_replicate_dEfret_median_median'], c=color, alpha=0.5, s=200, edgecolor='k', linewidth=2, zorder=2)
        cur_ax.errorbar(new_plotting_df['timepoint'], new_plotting_df['biological_replicate_dEfret_median_median'],yerr=new_plotting_df['biological_replicate_dEfret_median_std'], color='k', linewidth=0, elinewidth=2, zorder=3)
        cur_ax.plot(new_plotting_df['timepoint'], new_plotting_df['biological_replicate_dEfret_median_median'], zorder=2, c=color)

    construct = tup[0][0]
    cur_ax.set_title("T"+str(construct))
    cur_ax.axvline(2, color='k', linestyle='--',zorder=1)
    cur_ax.axhline(0.0, color='k', linestyle='--', zorder=1)
    cur_ax.set_xticks([0, 2, 4, 8])
    cur_ax.set_xticklabels([0, 12, 24, 48])

fig.supylabel(delta+"E$_f$")
fig.supxlabel("HPI")
plt.subplots_adjust(hspace=0.4, wspace=0.1)
#plt.savefig(r'W:/OneDrive/Virus/Paper1/new_fig/timelineplot_infection.svg', format='svg', bbox_inches='tight')
```
